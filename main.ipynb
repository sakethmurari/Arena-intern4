```python
import pandas as pd
import matplotlib.pyplot as plt
import os


# Load data
df = pd.read_csv("data/sales_data2.csv")

# Data Cleaning
df.isnull().sum()
df.drop_duplicates(inplace=True)

# Data Analysis

total_sales = df["Total_Sales"].sum()
best_product = df.groupby("Product")["Total_Sales"].sum().idxmax()
print(f"Total Sales: ₹{total_sales}")
print(f"Best Selling Product: ₹{best_product}")

```

```python
# -------------------------
# Bar Chart: Sales by Product
# -------------------------
product_sales = df.groupby("Product")["Total_Sales"].sum()

plt.figure()
product_sales.plot(kind="bar")
plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.savefig("visualizations/sales_by_product.png")
plt.show()
```

```python
# -------------------------
# Line Chart: Sales Trend by Product
# -------------------------
product_sales = df.groupby("Product")["Total_Sales"].sum()

plt.figure()
product_sales.plot(kind="line", marker="o")
plt.title("Sales Trend by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.savefig("visualizations/product_sales_trend.png")
plt.show()
```

```python
plt.figure()
product_sales.plot(kind="pie", autopct="%1.1f%%")
plt.title("Sales Distribution by Product")
plt.ylabel("")
plt.axis("equal")
plt.tight_layout()

plt.savefig("visualizations/expense_distribution.png")
plt.show()

print("Charts created successfully.")
```
